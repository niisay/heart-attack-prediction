![image.png](https://i.imgur.com/SU2V1XA.jpeg)

* Kalp krizi (miyokard enfarktüsü), kalbi besleyen koroner arterlerin tıkanması sonucu kalp kasının oksijensiz kalmasıyla ortaya çıkan hayati tehlike oluşturan bir durumdur. Bu durum kalp dokusunda kalıcı hasara yol açabilir.
* Kalp krizi genellikle göğüs bölgesinde baskı, sıkışma ve ağrı ile kendini gösterir. Erken teşhis ve hızlı müdahale, kalp kası hasarını en aza indirmek açısından kritik öneme sahiptir.

## 📁 Veri İçeriği 

Bu projede kullanılan veri seti, bireylerin sağlık bilgilerini içermekte olup kalp krizi riskini tahmin etmeye yönelik çeşitli özellikler barındırmaktadır. Aşağıda veri setinde bulunan değişkenler açıklanmıştır:

---

### 🔢 Demografik Özellikler
- **age**: Hastanın yaşı  
- **sex**: Hastanın cinsiyeti (1 = erkek, 0 = kadın)  

---

### ❤️ Klinik ve Semptom Bilgileri
- **cp (Chest Pain)**: Göğüs ağrısı türü  
  - 1: Tipik anjina  
  - 2: Atipik anjina  
  - 3: Anjina dışı ağrı  
  - 4: Belirti yok (asemptomatik)  

- **trestbps**: Dinlenme halindeki kan basıncı (mmHg)  
- **chol**: Kolesterol seviyesi (mg/dl)  
- **fbs**: Açlık kan şekeri > 120 mg/dl (1 = evet, 0 = hayır)  

---

### 🫀 Kalp ile İlgili Ölçümler
- **restecg**: Dinlenme elektrokardiyografi sonuçları  
  - 0: Normal  
  - 1: ST-T dalga anormalliği  
  - 2: Sol ventrikül hipertrofisi  

- **thalach**: Ulaşılan maksimum kalp atış hızı  
- **exang**: Egzersize bağlı anjina (1 = evet, 0 = hayır)  
- **oldpeak**: Egzersize bağlı ST depresyonu  

---

### 📈 Ek Klinik Göstergeler
- **slope**: ST segmentinin eğimi  
  - 1: Yukarı eğimli  
  - 2: Düz  
  - 3: Aşağı eğimli  

- **ca**: Boyanmış büyük damar sayısı (0-3)  
- **thal**: Talasemi durumu  
  - 3: Normal  
  - 6: Sabit kusur  
  - 7: Geri döndürülebilir kusur  

---

### 🎯 Hedef Değişken
- **target**:  
  - 0 → Kalp krizi geçirme olasılığı düşük  
  - 1 → Kalp krizi geçirme olasılığı yüksek  

---

## 🏷️ Kategorik Özellik Analizi 

Bu bölümde kategorik değişkenlerin hedef değişken (target) ile ilişkisi incelenmiştir. Grafikler üzerinden elde edilen gözlemler aşağıda sunulmuştur:

---

### 📊 Önemli Gözlemler

- **sex (cinsiyet)**:  
  Erkek bireylerin (1) sayısı daha fazla olmasına rağmen, kalp krizi geçiren bireylerin her iki cinsiyette de bulunduğu görülmektedir. Ancak erkeklerde riskin biraz daha yüksek olduğu söylenebilir.

- **cp (göğüs ağrısı türü)**:  
  Belirli göğüs ağrısı türlerinde (özellikle 2 ve 3 değerlerinde) kalp krizi geçiren bireylerin sayısının daha fazla olduğu dikkat çekmektedir. Bu değişkenin güçlü bir belirleyici olduğu söylenebilir.

- **fbs (açlık kan şekeri)**:  
  Açlık kan şekeri yüksek olan (1) bireylerin sayısı azdır ve bu değişkenin kalp krizi ile güçlü bir ilişki göstermediği gözlemlenmiştir.

- **restecg (EKG sonuçları)**:  
  Özellikle 1 değerine sahip bireylerde kalp krizi vakalarının daha fazla olduğu görülmektedir.

- **exang (egzersize bağlı anjina)**:

- **slope (ST segment eğimi)**:  
  2 değerine sahip bireylerde kalp krizi vakalarının daha fazla olduğu görülmektedir. Bu da slope değişkeninin önemli olduğunu göstermektedir.

- **ca (damar sayısı)**:  
  Damar sayısı arttıkça kalp krizi riski değişmektedir. Özellikle düşük damar sayılarında bile riskin bulunduğu dikkat çekmektedir.

- **thal**:  
  Özellikle 2 değerine sahip bireylerde kalp krizi oranının daha yüksek olduğu görülmektedir.

- **target (hedef değişken)**:

## ⚖️ Standardizasyon 

Bu bölümde sayısal değişkenler üzerinde standardizasyon (ölçekleme) işlemi uygulanmıştır.

Makine öğrenmesi modellerinde farklı ölçeklerdeki veriler model performansını olumsuz etkileyebilir. Örneğin bazı değişkenler büyük değerler alırken bazıları küçük aralıklarda olabilir. Bu durum modelin bazı değişkenlere daha fazla ağırlık vermesine neden olabilir.

Bu problemi önlemek için **StandardScaler** yöntemi kullanılmıştır.

---

### 📊 StandardScaler Nedir?

StandardScaler, verileri şu şekilde dönüştürür:

- Ortalama (mean) = 0  
- Standart sapma (std) = 1  

Yani her değişken aynı ölçeğe getirilir.

---

### 🔧 Uygulanan İşlem

Sayısal değişkenler üzerinde `fit_transform()` fonksiyonu kullanılarak standardizasyon işlemi gerçekleştirilmiştir.

Bu işlem sonucunda:
- Veri ortalaması sıfıra yakın  
- Standart sapması bire yakın  

hale getirilmiştir.

---

### 💡 Neden Önemli?

Standardizasyon özellikle şu modeller için kritik öneme sahiptir:

- Lojistik Regresyon  
- KNN (K-Nearest Neighbors)  
- SVM (Support Vector Machine)  

Bu modeller mesafe tabanlı çalıştığı için ölçek farklarından etkilenir.

---

### 📌 Genel Değerlendirme

Standardizasyon işlemi sayesinde modelin tüm değişkenleri eşit şekilde değerlendirmesi sağlanmış ve daha doğru sonuçlar elde edilmesi hedeflenmiştir.

```python
scaler = StandardScaler()
scaler
```

```python
scaled_array = scaler.fit_transform(df[numeric_list[:-1]])
```

```python
scaled_df = pd.DataFrame(scaled_array, columns=numeric_list[:-1])
scaled_df.head()
```

<a id="aykiri"></a>
## 🚨 Aykırı Değer Algılama 

Bu bölümde sayısal değişkenlerde bulunan aykırı değerler (outliers) IQR (Interquartile Range) yöntemi kullanılarak tespit edilmiştir.

---

### 📊 IQR Yöntemi Nedir?

IQR yöntemi, verinin orta %50’lik kısmını temel alır:

- Q1 (25. yüzdelik dilim)
- Q3 (75. yüzdelik dilim)
- IQR = Q3 - Q1

Aykırı değerler şu şekilde belirlenir:

- Alt sınır = Q1 - 1.5 × IQR  
- Üst sınır = Q3 + 1.5 × IQR  

Bu sınırların dışında kalan değerler aykırı değer olarak kabul edilir.

---

### 🔍 Analiz Sonuçları

Yapılan analiz sonucunda özellikle bazı değişkenlerde aykırı değerler tespit edilmiştir:

- **chol (kolesterol)** → yüksek değerlerde aykırılıklar  
- **oldpeak** → geniş dağılım ve uç değerler  
- **trestbps** → sınırlı sayıda aykırı değer  

Bu aykırı değerler veri setinden çıkarılmış ve veri boyutunda azalma gözlemlenmiştir.

---

### 📉 Veri Boyutundaki Değişim

Aykırı değerlerin temizlenmesi sonucunda veri setindeki gözlem sayısında azalma olmuştur. Bu durum, veri kalitesini artırmak amacıyla yapılmıştır.

---

### 💡 Genel Değerlendirme

Aykırı değerlerin temizlenmesi:

- Model performansını artırabilir  
- Gürültüyü (noise) azaltır  
- Daha doğru tahminler yapılmasını sağlar  

Ancak çok fazla veri silinmesi durumunda modelin genelleme yeteneği olumsuz etkilenebilir. Bu nedenle aykırı değer temizleme işlemi dikkatli uygulanmalıdır.

```python
numeric_list = ["age", "trestbps", "chol", "thalach", "oldpeak"]
df_numeric = df.loc[:, numeric_list]
df_numeric.head()
```

```python
df.describe()
```

```python
#  Aykırı Değer Algılama 

df_clean = df.copy()

for col in numeric_list:
    
    Q1 = np.percentile(df_clean[col], 25)
    Q3 = np.percentile(df_clean[col], 75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    print(f"{col} için sınırlar: Alt = {lower}, Üst = {upper}")

    # Filtreleme (drop yerine daha güvenli)
    before = df_clean.shape[0]

    df_clean = df_clean[(df_clean[col] >= lower) & (df_clean[col] <= upper)]

    after = df_clean.shape[0]

    print(f"{col} için silinen satır sayısı: {before - after}")
    print("Yeni shape:", df_clean.shape)
    print("-" * 40)

# Son veri boyutu
df_clean.shape
```

```python
lr_searcher.fit(x_train, y_train)
```

```python
print("Best Parameters: ",lr_searcher.best_params_)
```

```python
y_pred = lr_searcher.predict(x_test)
```

```python
print("Test accuracy: {}".format(accuracy_score(y_test, y_pred)))
```
